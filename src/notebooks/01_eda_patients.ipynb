```python
from pathlib import Path
import sys

# Retrouver la racine, que Jupyter soit lancé depuis
# la racine du projet ou depuis notebooks/.
current_dir = Path.cwd().resolve()

PROJECT_ROOT = next(
    (folder for folder in (current_dir, *current_dir.parents)
     if (folder / "src" / "config.py").is_file()),
    None,
)

if PROJECT_ROOT is None:
    raise FileNotFoundError(
        "Impossible de trouver src/config.py. "
        "Ouvre Jupyter depuis le dossier du projet."
    )

# Permettre les imports depuis le projet.
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.config import RAW_DATA_DIR

print("Racine du projet :", PROJECT_ROOT)
print("Dossier des données :", RAW_DATA_DIR)
print("Le dossier existe :", RAW_DATA_DIR.is_dir())
```
